# Gambaran Alur Prediksi Percobaan Bunuh Diri dengan Binary Tree

```python
# Definisi Node (Simpul)
class Node:
    def __init__(self, data):
        self.data = data # nilai data               
        self.left = None # anak kiri            
        self.right = None # anak kanan               

# Membuat Node secara manual
root = Node("Try Sucide")
node1 = Node("X3 : Available")
node2 = Node("X3 : No Available")
node3 = Node("X6 : 0, 1, 11, 12, 2, 3, 8")
node4 = Node("X6 : 4, 5, 6, 7")
node5 = Node("X4 : 2, 3, 5, 7, 8")
node6 = Node("X4 : 1, 4, 6")
node7 = Node("X6 : 1, 11, 2")
node8 = Node("X6 : 0, 8")
node9 = Node("X5 : 0 / X2 : Adult, Teens, Unknown")
node10 = Node("X5 : 1 / X2 : Elderly, Kids, Young, Adult")

# Menghubungkan tiap node
root.left = node1
root.right = node2
node1.left = node3
node1.right = node4
node3.left = node5
node3.right = node6
node6.left = node7
node6.right = node8
node8.left = node9
node8.right = node10

# Fungsi untuk menampilkan Tree (Traversal Sederhana)
def cetak_tree(node, prefix="", is_left=True):
    if node is not None:
        # Menentukan simbol garis untuk node saat ini
        if prefix == "":
            print(str(node.data)) # Node Root tidak memakai garis pembuka
        else:
            connector = "├── " if is_left else "└── "
            print(prefix + connector + str(node.data))
        
        # Menambahkan garis vertikal untuk anak-anak node berikutnya
        new_prefix = prefix + ("│   " if is_left else "    ")
        
        # Rekursi untuk anak kiri dan kanan
        if node.left or node.right:
            cetak_tree(node.left, new_prefix, is_left=True)
            cetak_tree(node.right, new_prefix, is_left=False)
# Menjalankan fungsi
print("Struktur Tree:")
cetak_tree(root)
```

# Prediksi Ketidakberhasilan Percobaan Bunuh Diri dengan Binary Tree

```python
# Variabel

VARIABLES = {  # Dictionary utama untuk menyimpan semua variabel dan isi kategorinya

    "X1": {  # variabel X1
        "name": "Gender",  # Nama variabel
        "values": {  # Isi nilai yang diperbolehkan pada variabel
            1: "Man",  # Jika input 1 berarti laki-laki
            2: "Woman",  # Jika input 2 berarti perempuan
            0: "Unknown"  # Jika input 0 berarti tidak diketahui
        }
    },
    
    "X2": { 
        "name": "Age",  # Umuur
        "values": {
            0: "Unknown",  # Umur tidak diketahui
            1: "Children (<17 years old)",  # Anak-anak
            2: "Adolescent (17–19 years old)",  # Remaja
            3: "Young Adult (20–39 years old)",  # Dewasa muda
            4: "Mature Adult (40–59 years old)",  # Dewasa matang
            5: "Elderly (>=60 years old)"  # Lansia
        }
    },

    "X3": { 
        "name": "Rescue Action",  # Penyelamatan
        "values": {
            1: "Available",  # Ada tindakan penyelamatan
            2: "Not Available"  # Tidak ada tindakan penyelamatan
        }
    },

    "X4": { 
        "name": "Attempt Method",  # Metode percobaan bunuh diri
        "values": {
            1: "Hanging Oneself",  # Gantung diri
            2: "Taking Drugs, Poisons, or Dangerous Chemicals",  # Racun/obat
            3: "Cutting the Neck or Abdomen with a Knife",  # Melukai dengan pisau
            4: "Burning Oneself",  # Membakar diri
            5: "Jumping from a Bridge, Building, or Tall Structure",  # Lompat dari tempat tinggi
            6: "Jumping into a River, Lake, Pond, Sea, or Well",  # Lompat ke air
            7: "Crashing into a Train",  # Menabrak kereta
            8: "Other Methods"  # Metode lainnya
        }
    },

# variabel selai 0 & 1 pada x5 tidak dimasukkan karena agar tidak terjadi kesalahn pada input
    "X5": {  # Variabel X5
        "name": "Mental Disorder",  # Gangguan mental
        "values": {
            1: "Depression",  # Depresi
            0: "Unknown"  # Tidak diketahui
        }
    },

# variabel 9 & 10 pada x5 tidak dimasukkan karena agar tidak terjadi kesalahn pada input
# poin 12 kami gabung dengan 0, dan 3 dengan 2, hal ini agar tidak ada kesalahan output dengan mempertimbangkan kemiripan faktor
    "X6": {  
        "name": "Trigger Risk",  # Faktor pemicu risiko
        "values": {
            0: "Unknown or (12) Other Triggers",  # Tidak diketahui
            1: "Debt, Online Loans, or Financial Problems",  # Masalah keuangan
            2: "Breakup, Infidelity, or Romantic Problems, (3) Divorce",  # Masalah percintaan
            4: "Killing a Spouse or Family Member",  # Membunuh pasangan/keluarga
            5: "Being a Criminal Suspect",  # Tersangka kriminal
            6: "Prolonged Illness",  # Penyakit berkepanjangan
            7: "Death in the Family",  # Kematian keluarga
            8: "Job Loss (Layoff)",  # Kehilangan pekerjaan
            11: "Conflict with Family, Spouse, or Children"  # Konflik keluarga
        }
    }
}

# Fungsi input
def input_variable(var_code): # Membuat fungsi dengan parameter var_code

    values = VARIABLES[var_code]["values"] # Mengambil dictionary "values" dari VARIABLES berdasarkan var_code

    print("\n_________________________________")
    print(f"{var_code} - {VARIABLES[var_code]['name']}") # Menampilkan kode variabel dan nama variabel
    print("_________________________________")
    for k, v in values.items(): # Melakukan loop pada dictionary values & .items() mengambil pasangan key dan value
        print(f"{k} = {v}") # Menampilkan setiap pilihan

    while True: # Perulangan
        try: # if eror to except
            pilihan = int(input("\nMasukkan pilihan: ")) # Meminta input dari user

            if pilihan in values: #cek pilihan apakah ada di values
                return pilihan # Kalau sesuai maka kembalikan nilai

            print("Pilihan tidak tersedia") # jika tidak ada pada daftar pilihan

        except ValueError:
            print("Input harus angka") # jika input bukan angka

# Fungsi Binary Tree
def suicide_tree(X3, X4, X6, X2=None, X5=None):  # Fungsi utama Binary

# Anggap rootnya Try Suicide
    if X3 == 2:  # Jika tidak ada rescue action
        return {  # Mengembalikan hasil prediksi
            "prediction": "Suicide",  # Prediksi akhir
            "probability": {  # Probabilitas masing-masing kelas
                "Failed Suicide": 0.02,
                "Suicide": 0.98
            },
            "node_percentage": "71%"  # Persentase data pada node tersebut
        }
    elif X3 == 1:  # Jika ada rescue action
        if X6 in [4, 5, 6, 7]:  # Jika trigger termasuk kategori berat
            return {
                "prediction": "Suicide",
                "probability": {
                    "Failed Suicide": 0.17,
                    "Suicide": 0.83
                },
                "node_percentage": "4%"
            }
        elif X6 in [0, 1, 11, 12, 2, 3, 8]:  # Trigger kategori lain
            if X4 in [2, 3, 5, 7, 8]:  # Jika metode termasuk kelompok tertentu
                return {
                    "prediction": "Failed Suicide",
                    "probability": {
                        "Failed Suicide": 0.87,
                        "Suicide": 0.13
                    },
                    "node_percentage": "15%"
                }
            elif X4 in [1, 4, 6]:  # Jika metode termasuk kelompok lain
                if X6 in [1, 11, 2]:  # Trigger tertentu
                    return {
                        "prediction": "Failed Suicide",
                        "probability": {
                            "Failed Suicide": 0.86,
                            "Suicide": 0.14
                        },
                        "node_percentage": "2%"
                    }
                elif X6 in [0, 8]:  # Jika trigger unknown atau kehilangan pekerjaan
                    if X2 is not None:  # Jika user mengisi X2
                        if X2 in [4, 2, 0]:  # Kelompok umur tertentu
                            return {
                                "prediction": "Failed Suicide",
                                "probability": {
                                    "Failed Suicide": 0.54,
                                    "Suicide": 0.46
                                },
                                "node_percentage": "4%"
                            }
                        elif X2 in [5, 1, 3]:  # Kelompok umur lainnya
                            return {
                                "prediction": "Suicide",
                                "probability": {
                                    "Failed Suicide": 0.30,
                                    "Suicide": 0.70
                                },
                                "node_percentage": "3%"
                            }
                    elif X5 is not None:  # Jika user mengisi X5
                        if X5 == 0:  # Jika mental disorder unknown
                            return {
                                "prediction": "Failed Suicide",
                                "probability": {
                                    "Failed Suicide": 0.54,
                                    "Suicide": 0.46
                                },
                                "node_percentage": "4%"
                            }
                        elif X5 == 1:  # Jika depresi
                            return {
                                "prediction": "Suicide",
                                "probability": {
                                    "Failed Suicide": 0.30,
                                    "Suicide": 0.70
                                },
                                "node_percentage": "3%"
                            }
                        else:
                            return {
                                "message": "Nilai X5 tidak sesuai tree"  # Jika nilai X5 tidak cocok
                            }
                    else:

                        return {
                            "message": "Isi salah satu X2 atau X5"  # Jika X2 dan X5 kosong
                        }
    return {
        "message": "Kombinasi tidak sesuai tree"  # Jika kombinasi tidak ditemukan di tree
    }

# OP

print("\nSUICIDE PREDICTION")  

# Input utama
X3 = input_variable("X3")  # Input variabel X3
X6 = input_variable("X6")  # Input variabel X6
X4 = input_variable("X4")  # Input variabel X4

# Pilih X2 or X5
print("\n_________________________________")
print("Pilih input untuk variabel akhir")
print("_________________________________")
print("1 = X2 (Age)")  # Pilihan menggunakan umur
print("2 = X5 (Mental Disorder)")  # Pilihan menggunakan mental disorder

pilihan = input("\nPilih (1/2): ")  # Input pilihan user

if pilihan == "1":  # Jika memilih X2

    X2 = input_variable("X2")  # Input umur

    hasil = suicide_tree(  # Memanggil fungsi decision tree
        X3=X3,
        X4=X4,
        X6=X6,
        X2=X2
    )

elif pilihan == "2":  # Jika memilih X5

    X5 = input_variable("X5")  # Input mental disorder

    hasil = suicide_tree(  # Memanggil fungsi decision tree
        X3=X3,
        X4=X4,
        X6=X6,
        X5=X5
    )

else:

    hasil = {
        "message": "Pilihan harus 1 atau 2"  # Jika input selain 1 dan 2
    }

# uotput

print("\n_________________________________")
print("HASIL PREDIKSI")
print("_________________________________")

for key, value in hasil.items():  # Loop seluruh isi hasil prediksi
    print(f"{key} : {value}")  # Menampilkan hasil
```
