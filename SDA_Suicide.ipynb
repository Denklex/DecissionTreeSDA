# Gambaran Alur Prediksi Percobaan Bunuh Diri dengan Binary Tree

```python
# Definisi Node (Simpul)
class Node:
    def __init__(self, data):
        self.data = data # nilai data               
        self.left = None # anak kiri            
        self.right = None # anak kanan               

# Membuat Node secara manual
root = Node("Try Sucide")
node1 = Node("X3 : Available")
node2 = Node("X3 : No Available")
node3 = Node("X6 : 0, 1, 11, 12, 2, 3, 8")
node4 = Node("X6 : 4, 5, 6, 7")
node5 = Node("X4 : 2, 3, 5, 7, 8")
node6 = Node("X4 : 1, 4, 6")
node7 = Node("X6 : 1, 11, 2")
node8 = Node("X6 : 0, 8")
node9 = Node("X5 : 0 / X2 : Adult, Teens, Unknown")
node10 = Node("X5 : 1 / X2 : Elderly, Kids, Young, Adult")

root.left = node1
root.right = node2
node1.left = node3
node1.right = node4
node3.left = node5
node3.right = node6
node6.left = node7
node6.right = node8
node8.left = node9
node8.right = node10

# Fungsi untuk menampilkan Tree (Traversal Sederhana)
def cetak_tree(node, prefix="", is_left=True):
    if node is not None:
        # Menentukan simbol garis untuk node saat ini
        if prefix == "":
            print(str(node.data)) # Node Root tidak memakai garis pembuka
        else:
            connector = "├── " if is_left else "└── "
            print(prefix + connector + str(node.data))
        
        # Menambahkan garis vertikal untuk anak-anak node berikutnya
        new_prefix = prefix + ("│   " if is_left else "    ")
        
        # Rekursi untuk anak kiri dan kanan
        if node.left or node.right:
            cetak_tree(node.left, new_prefix, is_left=True)
            cetak_tree(node.right, new_prefix, is_left=False)
# Menjalankan fungsi
print("Struktur Tree:")
cetak_tree(root)
```

# Prediksi Keberhasilan Percobaan Bunuh Diri dengan Binary Tree

```python
# Variabel

VARIABLES = {

    "X1": {
        "name": "Gender",
        "values": {
            1: "Man",
            2: "Woman",
            0: "Unknown"
        }
    },

    "X2": {
        "name": "Age",
        "values": {
            0: "Unknown",
            1: "Children (<17 years old)",
            2: "Adolescent (17–19 years old)",
            3: "Young Adult (20–39 years old)",
            4: "Mature Adult (40–59 years old)",
            5: "Elderly (>=60 years old)"
        }
    },

    "X3": {
        "name": "Rescue Action",
        "values": {
            1: "Available",
            2: "Not Available"
        }
    },

    "X4": {
        "name": "Attempt Method",
        "values": {
            1: "Hanging Oneself",
            2: "Taking Drugs, Poisons, or Dangerous Chemicals",
            3: "Cutting the Neck or Abdomen with a Knife",
            4: "Burning Oneself",
            5: "Jumping from a Bridge, Building, or Tall Structure",
            6: "Jumping into a River, Lake, Pond, Sea, or Well",
            7: "Crashing into a Train",
            8: "Other Methods",
            0: "Unknown"
        }
    },
# variabel selai 0 & 1 pada x5 tidak dimasukkan karena agar tidak terjadi kesalahn pada input
    "X5": {
        "name": "Mental Disorder",
        "values": {
            1: "Depression",
            0: "Unknown"
        }
    },

    "X6": {
        "name": "Trigger Risk",
        "values": {
            0: "Unknown",
            1: "Debt, Online Loans, or Financial Problems",
            2: "Breakup, Infidelity, or Romantic Problems",
            3: "Divorce",
            4: "Killing a Spouse or Family Member",
            5: "Being a Criminal Suspect",
            6: "Prolonged Illness",
            7: "Death in the Family",
            8: "Job Loss (Layoff)",
            9: "Problems at Work or Business Decline",
            10: "Gambling or Online Gambling",
            11: "Conflict with Family, Spouse, or Children",
            12: "Other Triggers"
        }
    }
}

# Fungsi input

def input_variable(var_code):

    variable = VARIABLES[var_code]

    print("\n_________________________________")
    print(f"{var_code} - {variable['name']}")
    print("_________________________________")

    for key, value in variable["values"].items():
        print(f"{key} = {value}")

    while True:

        try:

            pilihan = int(input(f"\nMasukkan {var_code}: "))

            if pilihan in variable["values"]:

                print(
                    f"Selected: {pilihan} "
                    f"({variable['values'][pilihan]})"
                )

                return pilihan

            else:
                print("Pilihan tidak tersedia")

        except ValueError:
            print("Input harus berupa angka")

# Fungsi Decision Tree

def suicide_tree(X3, X4, X6, X2=None, X5=None):

# Anggap rootnya Try Suicide
    if X3 == 2:

        return {
            "prediction": "Suicide",
            "probability": {
                "Failed Suicide": 0.02,
                "Suicide": 0.98
            },
            "node_percentage": "71%"
        }

    elif X3 == 1:

        if X6 in [4, 5, 6, 7]:

            return {
                "prediction": "Suicide",
                "probability": {
                    "Failed Suicide": 0.17,
                    "Suicide": 0.83
                },
                "node_percentage": "4%"
            }

        elif X6 in [0, 1, 11, 12, 2, 3, 8]:

            if X4 in [2, 3, 5, 7, 8]:

                return {
                    "prediction": "Failed Suicide",
                    "probability": {
                        "Failed Suicide": 0.87,
                        "Suicide": 0.13
                    },
                    "node_percentage": "15%"
                }
                
            elif X4 in [1, 4, 6]:

                if X6 in [1, 11, 2]:

                    return {
                        "prediction": "Failed Suicide",
                        "probability": {
                            "Failed Suicide": 0.86,
                            "Suicide": 0.14
                        },
                        "node_percentage": "2%"
                    }

                elif X6 in [0, 8]:

                    if X2 is not None:

                        if X2 in [4, 2, 0]:

                            return {
                                "prediction": "Failed Suicide",
                                "probability": {
                                    "Failed Suicide": 0.54,
                                    "Suicide": 0.46
                                },
                                "node_percentage": "4%"
                            }

                        elif X2 in [5, 1, 3]:

                            return {
                                "prediction": "Suicide",
                                "probability": {
                                    "Failed Suicide": 0.30,
                                    "Suicide": 0.70
                                },
                                "node_percentage": "3%"
                            }

                    elif X5 is not None:

                        if X5 == 0:

                            return {
                                "prediction": "Failed Suicide",
                                "probability": {
                                    "Failed Suicide": 0.54,
                                    "Suicide": 0.46
                                },
                                "node_percentage": "4%"
                            }

                        elif X5 == 1:

                            return {
                                "prediction": "Suicide",
                                "probability": {
                                    "Failed Suicide": 0.30,
                                    "Suicide": 0.70
                                },
                                "node_percentage": "3%"
                            }

                        else:

                            return {
                                "message": "Nilai X5 tidak sesuai tree"
                            }

                    else:

                        return {
                            "message": "Isi salah satu X2 atau X5"
                        }

    return {
        "message": "Kombinasi tidak sesuai tree"
    }

# OP

print("\nSUICIDE PREDICTION")

# Input utama
X3 = input_variable("X3")
X6 = input_variable("X6")
X4 = input_variable("X4")

# Pilih X2 or X5
print("\n_________________________________")
print("Pilih input untuk variabel akhir")
print("_________________________________")
print("1 = X2 (Age)")
print("2 = X5 (Mental Disorder)")

pilihan = input("\nPilih (1/2): ")

if pilihan == "1":

    X2 = input_variable("X2")

    hasil = suicide_tree(
        X3=X3,
        X4=X4,
        X6=X6,
        X2=X2
    )

elif pilihan == "2":

    X5 = input_variable("X5")

    hasil = suicide_tree(
        X3=X3,
        X4=X4,
        X6=X6,
        X5=X5
    )

else:

    hasil = {
        "message": "Pilihan harus 1 atau 2"
    }

# uotput

print("\n_________________________________")
print("HASIL PREDIKSI")
print("_________________________________")

for key, value in hasil.items():
    print(f"{key} : {value}")
```
